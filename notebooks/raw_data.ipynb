## 1. Core Dataset Schemas

The primary tables are compressed as numerical representations to ensure optimal processing speeds and storage footprint. 

### A. The Cases Table (`cases_yyyy.csv`)
This is the foundational lifecycle log for every lawsuit filed in the lower judiciary.

| Feature Name | Data Type | Description & Pipeline Utility |
| :--- | :--- | :--- |
| `ddl_case_id` | `string` | **Primary Key**. A unique, randomized identifier for every single court case. Used to link all lookup reference tables. |
| `year` | `integer` | The calendar year the lawsuit was officially filed in court. |
| `state_code` | `integer` | Numeric code indicating the state jurisdiction. Maps to `cases_state_key.csv`. |
| `dist_code` | `integer` | Numeric code indicating the district jurisdiction. Maps to `cases_district_key.csv`. |
| `court_no` | `integer` | Numeric code identifying the exact courtroom inside the designated district. Maps to `cases_court_key.csv`. |
| `cino` | `string` | **Case Information Number**. The official tracking identifier used inside the live government e-Courts web platform. |
| `judge_position` | `integer` | Numeric identifier representing the official rank or designation of the presiding judge. |
| `female_defendant` | `binary (0/1)`| LSTM name-classifier prediction indicating if the respondent/accused party is female. |
| `female_petitioner` | `binary (0/1)`| LSTM name-classifier prediction indicating if the plaintiff/petitioner is female. |
| `female_adv_def` | `binary (0/1)`| LSTM name-classifier prediction indicating if the defense lawyer is female. |
| `female_adv_pet` | `binary (0/1)`| LSTM name-classifier prediction indicating if the petitioner's lawyer is female. |
| `type_name` | `integer` | Numeric identifier representing the specific case classification. Maps to `type_name_key.csv`. |
| `purpose_name` | `integer` | Numeric identifier tracking the objective of a given scheduled listing. Maps to `purpose_name_key.csv`. |
| `disp_name` | `integer` | Numeric identifier indicating how the case wrapped up. Maps to `disp_name_key.csv`. |
| `date_of_filing` | `date` | The exact day the lawsuit was submitted to the court registry. Used to compute **Case Duration**. |
| `date_of_decision` | `date` | The exact day the case reached its final disposition. Used to compute **Case Duration**. |
| `date_first_list` | `date` | The procedural date indicating when the case first appeared on a judge's physical cause list. |
| `date_last_list` | `date` | The calendar date of the most recent historical hearing. |
| `date_next_list` | `date` | The upcoming scheduled hearing assignment date (null if the case is already resolved). |

### B. The Judges Profiles Table (`judges_clean.csv`)
This table acts as a log capturing judicial movements, seat transfers, and courtroom tenures.

| Feature Name | Data Type | Description & Pipeline Utility |
| :--- | :--- | :--- |
| `ddl_judge_id` | `string` | Unique identifier tracking individual judges across varying courtroom assignments. |
| `state_code` | `integer` | Numeric code mapping the state where the judge is actively deployed. |
| `dist_code` | `integer` | Numeric code mapping the specific judicial district of deployment. |
| `court_no` | `integer` | The designated courtroom number where the judge is presiding. |
| `judge_position` | `integer` | The designation ranking of the judge. Matches the `judge_position` field in the cases table. |
| `female_judge` | `binary (0/1)`| Name-classifier prediction indicating if the judge identifies as female. |
| `start_date` | `date` | The timestamp marking exactly when this judge assumed control of this specific courtroom seat. |
| `end_date` | `date` | The timestamp marking when the judge rotated out of this specific courtroom assignment. |

### C. The Acts & Sections Table (`acts_sections.csv`)
*Note: In specific versions of this public repository, this mapping dataset focuses heavily on criminal metadata rather than purely commercial act names.*

| Feature Name | Data Type | Description & Pipeline Utility |
| :--- | :--- | :--- |
| `ddl_case_id` | `string` | Foreign key mapping back to the primary case file records. |
| `act` | `integer` | Numeric identifier for the parent legislation. Maps to `act_key.csv`. |
| `section` | `integer` | Numeric identifier for the precise section/clause. Maps to `section_key.csv`. |
| `bailable_ipc` | `binary (0/1)`| Categorical flag tracking whether an IPC violation charge is a bailable offense. |
| `number_sections_ipc`| `integer` | Continuous count tracking total Indian Penal Code sections invoked under the case ID. |
| `criminal` | `binary (0/1)`| Target-helper flag highlighting if the case falls within a criminal court track. |

## 2. Reference Lookup Keys (The Translation Layers)

Because storing text variables millions of times across 81.2 million entries would cause extreme file bloating, the database isolates text names inside standalone lookup tables.

Your pipeline should import these files to interpret the raw code signals into clear categorical arrays or user-facing explanations:

* **`act_key.csv`**: Decodes `act` codes into literal text names (e.g., *Code of Civil Procedure, Negotiable Instruments Act*).
* **`section_key.csv`**: Decodes `section` codes into exact statutory definitions.
* **`type_name_key.csv`**: Decodes `type_name` identifiers into human-readable definitions (e.g., *Commercial Suit, Money Recovery Suit, Summary Suit*). **Mandatory step for filtering commercial scopes.**
* **`disp_name_key.csv`**: Translates final outcomes (e.g., *Dismissed, Settled, Allowed, Withdrawn*). **Mandatory step for training the Litigate vs. Settle logic**.
* **`purpose_name_key.csv`**: Translates the structural purpose behind active hearings (e.g., *For Evidence, For Arguments, For Admission*).
* **`cases_state_key.csv`**: Translates `state_code` into regional strings (e.g., *Maharashtra, Delhi, Karnataka*).
* **`cases_district_key.csv`**: Translates `dist_code` into localized district geographic names.
* **`cases_court_key.csv`**: Translates `court_no` into descriptive courtroom infrastructure groupings.

```python
target_commercial_codes = set(commercial_rows['type_name'].unique())
```

```python
strict_whitelist = [
    'os', 'o.s.', 'xxxx--o s',
    'money suit h a', 'money suit  t',
    'original suit/hm',
    'original (title/partition/money) suit',
    'suit for mandatory injunction(civil)',
    'suit for cancellation of agreement',
    'suit for recovery',
    'summary suit', 'sum',
    'stc',  # summons to case — civil
    'miscellaneous civil',
    'money suit',
]

# Get the type_name codes for these
strict_codes = type_key[
    type_key['type_name_s'].str.strip().str.lower().isin([x.lower() for x in strict_whitelist])
]['type_name'].unique()

print(f"Strict codes: {len(strict_codes)}")
```

```python
import pandas as pd

# 1. Your combined list of commercial and cheque bounce numeric codes
# (Extracted from your type_name_key filtering steps)
commercial_codes_set = set(strict_codes)  # Start with the strict whitelist codes

case_files = [
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2010.csv",
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2011.csv",
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2012.csv",
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2013.csv",
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2014.csv",
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2015.csv",
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2016.csv",
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2017.csv",
    r"C:\Users\Karnaveer Singh\Justice_Hq\csv\cases\cases_2018.csv",
]

filtered_slices = []

print("--- Starting Dynamic Commercial Extraction Pipeline ---")
for path in case_files:
    # Extract year string from path for clean logging
    year_label = path.split("_")[-1].replace(".csv", "")
    print(f"Streaming and filtering cases for Year: {year_label}...")
    
    # Read the file one year at a time to protect laptop RAM
    year_df = pd.read_csv(path)
    
    # Single, bulletproof filter: Grab only the target commercial rows
    commercial_slice = year_df[year_df['type_name'].isin(commercial_codes_set)].copy()
    
    # Save the compressed slice and immediately wipe the heavy raw dataframe from memory
    filtered_slices.append(commercial_slice)
    del year_df

# Concatenate only the clean commercial files together
comm_cases = pd.concat(filtered_slices, ignore_index=True)

print(f"\n[SUCCESS] Pipeline Complete!")
print(f"Total isolated commercial rows ready for ML modeling: {len(comm_cases):,}")
```

```python
# Convert dates
df['date_of_filing'] = pd.to_datetime(df['date_of_filing'], errors='coerce')
df['date_of_decision'] = pd.to_datetime(df['date_of_decision'], errors='coerce')

# Calculate resolution time in days
resolved = df[df['date_of_decision'].notna()].copy()
resolved['resolution_days'] = (resolved['date_of_decision'] - resolved['date_of_filing']).dt.days

# Drop negative or zero values (data errors)
resolved = resolved[resolved['resolution_days'] > 0]

print(f"Resolved cases: {len(resolved):,}")
print(f"\nResolution time (days):")
print(resolved['resolution_days'].describe())
print(f"\nMedian resolution: {resolved['resolution_days'].median():.0f} days ({resolved['resolution_days'].median()/365:.1f} years)")
```

```python
# Diagnose the tail


# this is error in data entry found out by diagnosis
print("Distribution of extreme values:")
print(f"Cases over 10 years (3650 days): {(resolved['resolution_days'] > 3650).sum():,}")
print(f"Cases over 20 years (7300 days): {(resolved['resolution_days'] > 7300).sum():,}")
print(f"Cases over 50 years (18250 days): {(resolved['resolution_days'] > 18250).sum():,}")

# What are the filing dates of these extreme cases?
extreme = resolved[resolved['resolution_days'] > 18250]
print(f"\nExtreme cases filing dates:")
print(extreme['date_of_filing'].dt.year.value_counts())
print(f"\nExtreme cases decision dates:")
print(extreme['date_of_decision'].dt.year.value_counts())
```

```python
# Cap at 30 years - anything beyond is a data entry error
resolved_clean = resolved[resolved['resolution_days'] <= 10950].copy()

print(f"Dropped: {len(resolved) - len(resolved_clean):,} corrupt cases")
print(f"Clean resolved cases: {len(resolved_clean):,}")
print(f"\nFinal resolution time:")
print(f"Median: {resolved_clean['resolution_days'].median():.0f} days ({resolved_clean['resolution_days'].median()/365:.1f} years)")
print(f"Mean: {resolved_clean['resolution_days'].mean():.0f} days")
print(f"90th percentile: {resolved_clean['resolution_days'].quantile(.90):.0f} days ({resolved_clean['resolution_days'].quantile(.90)/365:.1f} years)")

# Target variable
def resolution_bucket(days):
    if days <= 180:
        return '0_under_6months'
    elif days <= 730:
        return '1_six_to_24months'
    else:
        return '2_over_2years'

resolved_clean['resolution_bucket'] = resolved_clean['resolution_days'].apply(resolution_bucket)

print("\nTarget variable distribution:")
dist = resolved_clean['resolution_bucket'].value_counts().sort_index()
pct = resolved_clean['resolution_bucket'].value_counts(normalize=True).mul(100).round(1).sort_index()
for bucket, count, p in zip(dist.index, dist.values, pct.values):
    print(f"  {bucket}: {count:,} cases ({p}%)")
```
